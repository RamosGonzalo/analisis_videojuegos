```python
ventas_por_generos = df.groupby('genero')['ventas_global'].sum().sort_values(ascending=False)
```

```python
ventas_por_generos.plot(kind='bar')
```

```python
ventas_por_generos.plot(kind='bar')
plt.title('Ventas Globales por Género')
plt.ylabel('Millones de copias')
plt.xlabel('Género')
plt.show()
```

```python
ventas_por_generos.plot(kind='barh')
plt.title('Ventas Globales por Género')
plt.xlabel('Millones de copias')
plt.ylabel('Género')
plt.show()
```

```python
ventas_por_generos.plot(kind='pie')
plt.title('Ventas Globales por Género')
plt.show()
```

```python
ventas_por_generos.plot(kind='line')
plt.title('Ventas Globales por Género')
plt.ylabel('Millones de copias')
plt.xlabel('Género')
plt.show()
```
